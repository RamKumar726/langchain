# **STM manually using messages** 

```python
from langchain_community.tools import DuckDuckGoSearchRun
from langchain.agents import create_agent
websearch_tool = DuckDuckGoSearchRun()

agent = create_agent(
    model="groq:openai/gpt-oss-120b",
    tools=[websearch_tool]
)
messages = []
while(True):

    user_input = input("User > ")
    if user_input == "exit":
        print("Bye.........")
        break
    human = {"role" : "user", "content": user_input}
    messages.append(human)
    response  = agent.invoke({"messages" : messages})
    assistant = {"role" : "assistant", "content": response['messages'][-1].content}
    messages.append(assistant)
    print(f"AI >  {response['messages'][-1].content}")


```

# **STM using langchain**

```python
from genericpath import exists
from langgraph.checkpoint.memory import InMemorySaver

agent_with_stm = create_agent(
    model="groq:openai/gpt-oss-120b",
    tools = [websearch_tool],
    checkpointer=InMemorySaver()
)

config = {
    "configurable":{
        "thread_id" : "user-1"
    }
}

while True:
    user_input = input("User > ")
    if(user_input == 'exit'):
        print("Bye Bye...!")
        break
    human = {"role" : "user", "content": user_input}
   
    response = agent_with_stm.invoke({"messages": human}, config)
    print("AI > ", response['messages'][-1].content)
```
